```python
import pandas as pd

# Nomes dos ficheiros
input_file = 'tmy_40.593_-8.585_2005_2023.csv'
output_file = 'tmy_2023_normalized.csv'

# Ler todas as linhas do ficheiro original
with open(input_file, 'r') as f:
    lines = f.readlines()

# Separar metadados e cabeçalho (linhas 0 a 17) dos dados (linha 18 em diante)
metadata = lines[:17]
data_header = lines[17]
data_rows = lines[18:]

processed_data = []
for row in data_rows:
    if row.strip():
        # O formato original é YYYYMMDD:HHMM
        # Substituímos os primeiros 4 caracteres por '2023'
        
        # Verificação de segurança para 29 de Fevereiro (2023 não é bissexto)
        if "0229:" in row:
            continue
            
        new_row = "2023" + row[4:]
        processed_data.append(new_row)

# Gravar o novo ficheiro mantendo a estrutura original
with open(output_file, 'w') as f:
    f.writelines(metadata)
    f.write(data_header)
    f.writelines(processed_data)

print(f"Sucesso! O ficheiro '{output_file}' foi criado.")
```
